# Importing Libraries

```python
#Importing the libraries 
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import missingno as msno
```

# Loading The Dataset

```python
#Loading the Dataset
data = pd.read_csv("C:\\Users\\Dell\\Downloads\\Titanic-Dataset.csv")
```

# Basic Exploration

```python
#Basic exploration
print(data.head())
print(data.info())
print(data.describe())
```

# Data Cleaning

```python
#Data Cleaning
#Checking for missing values
print(data.isnull().sum())
```

# Handling The Missing values

```python
#Handling missing values
data['Age'].fillna(data['Age'].median())
data['Embarked'].fillna(data['Embarked'].mode()[0])
```

# Visualizing Missing Values 

```python
#Visualizing missing values
msno.matrix(data)
plt.show()
```

**Observation:**  
- Most features have complete data.
- Some missing values might be present in 'Age' and 'Embarked' (if visible).
- Overall, the dataset is fairly clean.


# Univariate Analysis.

In this section, we will explore individual variables to understand their distributions and characteristics.

Exploring the distribution of passenger ages.


```python
#Univariate Analysis
#Histogram for Age Distribution
sns.histplot(data['Age'], kde= True)
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

**Observation:**

- Most passengers were between 20 and 40 years old.

- Very few passengers were under 10 or over 60, showing a young adult-heavy passenger list.

```python
#Countplot
#Survival Count
sns.countplot(x= 'Survived', data= data)
plt.title('Survival Count')
plt.xlabel('Survived')
plt.ylabel('Frequency')
plt.show()
```

**Observation:**

- Around 38% of passengers survived, while 62% did not.

- The dataset is slightly imbalanced toward non-survivors.

# Bivariate Analysis

```python
#Bivariate Analysis
#Bar Plot
#Gender by Survival
sns.barplot(x= 'Sex', y='Survived', data=data)
plt.title('Survival Rate By Gender')
plt.xlabel('Sex')
plt.ylabel('Survival Rate')
plt.show()
```

**Observation:**

- A much higher percentage of females survived compared to males.

- Gender was a strong factor influencing survival chances on the Titanic.

```python
#Bar Plot2 for Survival by Passenger Class
sns.barplot(x='Pclass', y= 'Survived', data=data)
plt.title('Survival Rate By Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate')
plt.show()
```

**Observation:**

- Passengers from 1st Class had significantly higher survival rates.

- Lower class passengers (especially 3rd class) faced lower survival chances.

```python
#Scatterplot for Fare Distribution
sns.scatterplot(x='Fare', y='Age', hue= 'Survived', data=data)
plt.title('Fare vs Age colored by Survival')
plt.show()
```

**Observation:**

- Passengers who paid higher fares generally had better survival chances.

- Younger passengers had varying survival rates across different fare amounts


# Multivariate Analysis

```python
#Multivariate Analysis

#Pairplot
sns.pairplot(data[['Survived','Age', 'Fare', 'Pclass']])
plt.show()

numeric_data= data.select_dtypes(include=['float64', 'int64'])
```

**Observation:**
- Clear visual separation is seen where higher class and higher fare passengers survived more.

- Younger first-class passengers especially had a better survival chance.

```python
#Heatmap
plt.figure(figsize=(10,6))
sns.heatmap(numeric_data.corr(),annot= True, cmap= 'coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

**Obsevation:**

- Fare and Pclass are strongly correlated with survival.

- Age and survival have a weaker correlation compared to class and fare.

# Advanced Visualizations 

```python
#Violinplot: Age distribution by Survival
sns.violinplot(x='Survived', y='Age', data=data)
plt.title('Age Distribution by Survival')
plt.xlabel('Survived')
plt.ylabel('Age')
plt.show()
```

**Observation:**

Younger passengers had a higher chance of survival.

Non-survivors were spread across all age groups, while survivors mostly were younger.

```python
# Swarmplot: Fare by Gender
sns.swarmplot(x='Sex', y='Fare', data=data)
plt.title('Fare paid by Gender')
plt.xlabel('Gender')
plt.ylabel('Fare')
plt.show()
```

**Observation:**

Males generally paid slightly lower fares than females.

A few outliers paid extremely high fares regardless of gender.

## Key Insights

- Most passengers who survived were in the age group of 20-40.
- Females had a much higher survival rate than males.
- Higher class passengers (1st class) had better survival chances.
- Families with more than 3 members were less likely to survive.
- Fare distribution is highly right-skewed, with a few paying exceptionally high fares.


# Conclusion

The exploratory data analysis revealed important factors influencing survival rates on the Titanic.
